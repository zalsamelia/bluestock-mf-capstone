```python
# ==========================================================
# Bluestock Mutual Fund Analytics
# Day 3 - Exploratory Data Analysis
# ==========================================================

# ==========================================================
# Import Libraries
# ==========================================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from pathlib import Path

# Display Options
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)
pd.set_option("display.max_colwidth", None)

print("=" * 60)
print("Libraries Successfully Imported")
print("=" * 60)

# ==========================================================
# Project Paths
# ==========================================================

PROJECT_ROOT = Path.cwd().parent

DATA_PATH = PROJECT_ROOT / "data" / "processed"

FIGURE_PATH = Path.cwd() / "figures"

FIGURE_PATH.mkdir(
    parents=True,
    exist_ok=True
)

print("=" * 60)
print("Project Paths")
print("=" * 60)

print(f"Project Root : {PROJECT_ROOT}")
print(f"Processed Data : {DATA_PATH}")
print(f"Figures Folder : {FIGURE_PATH}")

# ==========================================================
# Load Clean Datasets
# ==========================================================

fund_df = pd.read_csv(
    DATA_PATH / "clean_fund_master.csv",
    parse_dates=["launch_date"]
)

nav_df = pd.read_csv(
    DATA_PATH / "clean_nav_history.csv",
    parse_dates=["date"]
)

aum_df = pd.read_csv(
    DATA_PATH / "clean_aum_by_fund_house.csv",
    parse_dates=["date"]
)

sip_df = pd.read_csv(
    DATA_PATH / "clean_monthly_sip_inflows.csv",
    parse_dates=["month"]
)

category_df = pd.read_csv(
    DATA_PATH / "clean_category_inflows.csv",
    parse_dates=["month"]
)

folio_df = pd.read_csv(
    DATA_PATH / "clean_industry_folio_count.csv",
    parse_dates=["month"]
)

performance_df = pd.read_csv(
    DATA_PATH / "clean_scheme_performance.csv"
)

transaction_df = pd.read_csv(
    DATA_PATH / "clean_investor_transactions.csv",
    parse_dates=["transaction_date"]
)

portfolio_df = pd.read_csv(
    DATA_PATH / "clean_portfolio_holdings.csv",
    parse_dates=["portfolio_date"]
)

benchmark_df = pd.read_csv(
    DATA_PATH / "clean_benchmark_indices.csv",
    parse_dates=["date"]
)

print("=" * 60)
print("All datasets loaded successfully.")
print("=" * 60)

# ==========================================================
# Dataset Collection
# ==========================================================

datasets = {

    "Fund Master": fund_df,

    "NAV History": nav_df,

    "AUM by Fund House": aum_df,

    "Monthly SIP": sip_df,

    "Category Inflows": category_df,

    "Industry Folio": folio_df,

    "Scheme Performance": performance_df,

    "Investor Transactions": transaction_df,

    "Portfolio Holdings": portfolio_df,

    "Benchmark": benchmark_df

}

print("=" * 60)
print("Dataset Dictionary Created")
print("=" * 60)

print(f"Total Dataset : {len(datasets)}")

# ==========================================================
# Dataset Overview
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    print(f"Shape : {df.shape}")

    print("\nFirst Five Rows")

    display(df.head())

    # ==========================================================
# Dataset Dimensions
# ==========================================================

summary = []

for name, df in datasets.items():

    summary.append({

        "Dataset": name,

        "Rows": df.shape[0],

        "Columns": df.shape[1]

    })

summary_df = pd.DataFrame(summary)

summary_df

# ==========================================================
# Data Types
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    print(df.dtypes)

    # ==========================================================
# Dataset Information
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    df.info()

    print("\n")

    # ==========================================================
# Dataset Information
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    df.info()

    print("\n")

    # ==========================================================
# Dataset Size Comparison
# ==========================================================

plt.figure(figsize=(12,6))

plt.bar(

    summary_df["Dataset"],

    summary_df["Rows"]

)

plt.title("Number of Rows per Dataset")

plt.xlabel("Dataset")

plt.ylabel("Rows")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "dataset_size.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Missing Values
# ==========================================================

print("=" * 60)
print("Missing Value Assessment")
print("=" * 60)

missing_summary = []

for name, df in datasets.items():

    total_missing = df.isnull().sum().sum()

    missing_summary.append({

        "Dataset": name,

        "Missing Values": total_missing

    })

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    print(df.isnull().sum())

missing_df = pd.DataFrame(missing_summary)

print("\n")

print("=" * 60)

print("Overall Missing Value Summary")

print("=" * 60)

display(missing_df)

# ==========================================================
# Duplicate Records
# ==========================================================

print("=" * 60)
print("Duplicate Record Assessment")
print("=" * 60)

duplicate_summary = []

for name, df in datasets.items():

    duplicate_rows = df.duplicated().sum()

    duplicate_summary.append({

        "Dataset": name,

        "Duplicate Rows": duplicate_rows

    })

    print(f"{name} : {duplicate_rows}")

duplicate_df = pd.DataFrame(duplicate_summary)

print("\n")

display(duplicate_df)

# ==========================================================
# Unique Values
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    print(df.nunique())

    # ==========================================================
# Numeric Columns
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    numeric_columns = df.select_dtypes(include="number").columns

    print(list(numeric_columns))

    # ==========================================================
# Numeric Columns
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    numeric_columns = df.select_dtypes(include="number").columns

    print(list(numeric_columns))

    # ==========================================================
# Date Columns
# ==========================================================

for name, df in datasets.items():

    print("\n")

    print("=" * 60)

    print(name)

    print("=" * 60)

    date_columns = df.select_dtypes(include="datetime").columns

    print(list(date_columns))

    # ==========================================================
# Memory Usage
# ==========================================================

memory_summary = []

for name, df in datasets.items():

    memory_summary.append({

        "Dataset": name,

        "Memory (MB)": round(

            df.memory_usage(deep=True).sum() / 1024 / 1024,

            2

        )

    })

memory_df = pd.DataFrame(memory_summary)

display(memory_df)

# ==========================================================
# Missing Value Chart
# ==========================================================

plt.figure(figsize=(10,5))

plt.bar(

    missing_df["Dataset"],

    missing_df["Missing Values"]

)

plt.title("Missing Values by Dataset")

plt.xlabel("Dataset")

plt.ylabel("Missing Values")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "missing_values.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Duplicate Records Chart
# ==========================================================

plt.figure(figsize=(10,5))

plt.bar(

    duplicate_df["Dataset"],

    duplicate_df["Duplicate Rows"]

)

plt.title("Duplicate Records by Dataset")

plt.xlabel("Dataset")

plt.ylabel("Duplicate Rows")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "duplicate_records.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Memory Usage Chart
# ==========================================================

plt.figure(figsize=(10,5))

plt.bar(

    memory_df["Dataset"],

    memory_df["Memory (MB)"]

)

plt.title("Dataset Memory Usage")

plt.xlabel("Dataset")

plt.ylabel("Memory (MB)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "memory_usage.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Data Quality Summary
# ==========================================================

print("=" * 60)
print("Data Quality Assessment Summary")
print("=" * 60)

print(f"Total Datasets : {len(datasets)}")

print(f"Total Missing Values : {missing_df['Missing Values'].sum()}")

print(f"Total Duplicate Rows : {duplicate_df['Duplicate Rows'].sum()}")

print("\nThe cleaned datasets show good overall data quality.")

print("Most datasets contain complete records with consistent")

print("data types and minimal quality issues.")

print("The datasets are suitable for exploratory data analysis.")

# ==========================================================
# FUND MASTER ANALYSIS
# ==========================================================

print("=" * 60)
print("FUND MASTER ANALYSIS")
print("=" * 60)

# ==========================================================
# Analysis 1
# Distribution of Mutual Fund Categories
# ==========================================================

print("\nDistribution of Mutual Fund Categories")

category_count = (
    fund_df["category"]
    .value_counts()
    .sort_values(ascending=False)
)

display(category_count)

plt.figure(figsize=(8,5))

plt.bar(
    category_count.index,
    category_count.values
)

plt.title("Distribution of Mutual Fund Categories")

plt.xlabel("Category")

plt.ylabel("Number of Funds")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "fund_category_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 2
# Top 5 Fund Houses
# ==========================================================

print("\nTop 5 Fund Houses")

top_fund_house = (
    fund_df["fund_house"]
    .value_counts()
    .head(5)
)

display(top_fund_house)

plt.figure(figsize=(8,5))

plt.bar(
    top_fund_house.index,
    top_fund_house.values
)

plt.title("Top 5 Fund Houses")

plt.xlabel("Fund House")

plt.ylabel("Number of Mutual Funds")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "top_5_fund_house.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 3
# Risk Category Distribution
# ==========================================================

print("\nRisk Category Distribution")

risk_distribution = (
    fund_df["risk_category"]
    .value_counts()
)

display(risk_distribution)

plt.figure(figsize=(8,5))

plt.bar(
    risk_distribution.index,
    risk_distribution.values
)

plt.title("Risk Category Distribution")

plt.xlabel("Risk Category")

plt.ylabel("Number of Funds")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "risk_category_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 4
# Launch Year Trend
# ==========================================================

print("\nLaunch Year Trend")

launch_year = (
    fund_df["launch_date"]
    .dt.year
    .value_counts()
    .sort_index()
)

display(launch_year)

plt.figure(figsize=(10,5))

plt.plot(
    launch_year.index,
    launch_year.values,
    marker="o"
)

plt.title("Mutual Fund Launch Trend")

plt.xlabel("Launch Year")

plt.ylabel("Number of Funds")

plt.grid(True)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "launch_year_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 5
# Expense Ratio Distribution
# ==========================================================

print("\nExpense Ratio Distribution")

plt.figure(figsize=(8,5))

plt.hist(
    fund_df["expense_ratio_pct"],
    bins=10
)

plt.title("Expense Ratio Distribution")

plt.xlabel("Expense Ratio (%)")

plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "expense_ratio_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# PART 3B - NAV & PERFORMANCE ANALYSIS
# ==========================================================

print("=" * 60)
print("NAV & PERFORMANCE ANALYSIS")
print("=" * 60)

# ==========================================================
# Analysis 1
# NAV Distribution
# ==========================================================

print("\nNAV Distribution")

plt.figure(figsize=(8,5))

plt.hist(
    nav_df["nav"],
    bins=30
)

plt.title("Distribution of NAV")

plt.xlabel("NAV")

plt.ylabel("Frequency")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "nav_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 2
# Average NAV by Fund (Top 5)
# ==========================================================

print("\nTop 5 Average NAV")

average_nav = (

    nav_df

    .groupby("amfi_code")["nav"]

    .mean()

    .reset_index()

)

average_nav = average_nav.merge(

    fund_df[["amfi_code","scheme_name"]],

    on="amfi_code"

)

average_nav = average_nav.sort_values(

    by="nav",

    ascending=False

).head(5)

display(average_nav)

plt.figure(figsize=(10,5))

plt.bar(

    average_nav["scheme_name"],

    average_nav["nav"]

)

plt.xticks(rotation=20)

plt.title("Top 5 Average NAV")

plt.xlabel("Scheme")

plt.ylabel("Average NAV")

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "top5_average_nav.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Analysis 3
# Average 3-Year Return (Top 5)
# ==========================================================

print("\nTop 5 Three-Year Return")

top_return = (

    performance_df

    .sort_values(

        by="return_3yr_pct",

        ascending=False

    )

    .head(5)

)

display(

    top_return[[

        "scheme_name",

        "return_3yr_pct"

    ]]

)

plt.figure(figsize=(10,5))

plt.bar(

    top_return["scheme_name"],

    top_return["return_3yr_pct"]

)

plt.xticks(rotation=20)

plt.title("Top 5 Three-Year Return")

plt.xlabel("Scheme")

plt.ylabel("Return (%)")

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "top5_return.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Analysis 4
# Top 5 Sharpe Ratio
# ==========================================================

print("\nTop 5 Sharpe Ratio")

top_sharpe = (

    performance_df

    .sort_values(

        by="sharpe_ratio",

        ascending=False

    )

    .head(5)

)

display(

    top_sharpe[[

        "scheme_name",

        "sharpe_ratio"

    ]]

)

plt.figure(figsize=(10,5))

plt.bar(

    top_sharpe["scheme_name"],

    top_sharpe["sharpe_ratio"]

)

plt.xticks(rotation=20)

plt.title("Top 5 Sharpe Ratio")

plt.xlabel("Scheme")

plt.ylabel("Sharpe Ratio")

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "top5_sharpe_ratio.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
# Analysis 5
# Top 5 AUM
# ==========================================================

print("\nTop 5 Assets Under Management")

top_aum = (

    performance_df

    .sort_values(

        by="aum_crore",

        ascending=False

    )

    .head(5)

)

display(

    top_aum[[

        "scheme_name",

        "aum_crore"

    ]]

)

plt.figure(figsize=(10,5))

plt.bar(

    top_aum["scheme_name"],

    top_aum["aum_crore"]

)

plt.xticks(rotation=20)

plt.title("Top 5 Assets Under Management")

plt.xlabel("Scheme")

plt.ylabel("AUM (Crore)")

plt.tight_layout()

plt.savefig(

    FIGURE_PATH / "top5_aum.png",

    dpi=300,

    bbox_inches="tight"

)

plt.show()

# ==========================================================
#  INVESTOR TRANSACTION ANALYSIS
# ==========================================================

print("=" * 60)
print("INVESTOR TRANSACTION ANALYSIS")
print("=" * 60)

# ==========================================================
# Analysis 1
# Transaction Type Distribution
# ==========================================================

print("\nTransaction Type Distribution")

transaction_type = (
    transaction_df["transaction_type"]
    .value_counts()
)

display(transaction_type)

plt.figure(figsize=(8,5))

plt.bar(
    transaction_type.index,
    transaction_type.values
)

plt.title("Transaction Type Distribution")

plt.xlabel("Transaction Type")

plt.ylabel("Number of Transactions")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "transaction_type_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 2
# Top 5 States
# ==========================================================

print("\nTop 5 States")

top_states = (
    transaction_df["state"]
    .value_counts()
    .head(5)
)

display(top_states)

plt.figure(figsize=(9,5))

plt.bar(
    top_states.index,
    top_states.values
)

plt.title("Top 5 States by Transactions")

plt.xlabel("State")

plt.ylabel("Transactions")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "top5_states.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 3
# Age Group Distribution
# ==========================================================

print("\nAge Group Distribution")

age_group = (
    transaction_df["age_group"]
    .value_counts()
)

display(age_group)

plt.figure(figsize=(8,5))

plt.bar(
    age_group.index,
    age_group.values
)

plt.title("Investor Age Group")

plt.xlabel("Age Group")

plt.ylabel("Number of Investors")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "age_group_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 4
# Payment Mode Distribution
# ==========================================================

print("\nPayment Mode Distribution")

payment = (
    transaction_df["payment_mode"]
    .value_counts()
)

display(payment)

plt.figure(figsize=(8,5))

plt.bar(
    payment.index,
    payment.values
)

plt.title("Payment Mode Distribution")

plt.xlabel("Payment Mode")

plt.ylabel("Transactions")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "payment_mode_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 5
# Average Investment by Age Group
# ==========================================================

print("\nAverage Investment by Age Group")

avg_investment = (
    transaction_df
    .groupby("age_group")["amount_inr"]
    .mean()
    .sort_values(ascending=False)
)

display(avg_investment)

plt.figure(figsize=(8,5))

plt.bar(
    avg_investment.index,
    avg_investment.values
)

plt.title("Average Investment by Age Group")

plt.xlabel("Age Group")

plt.ylabel("Average Investment (INR)")

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "average_investment_age_group.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# INDUSTRY & MARKET ANALYSIS
# ==========================================================

print("=" * 60)
print("INDUSTRY & MARKET ANALYSIS")
print("=" * 60)

# ==========================================================
# Analysis 1
# AUM Trend
# ==========================================================

print("\nAUM Trend")

aum_trend = (
    aum_df
    .groupby("date")["aum_crore"]
    .sum()
    .reset_index()
)

display(aum_trend.head())

plt.figure(figsize=(10,5))

plt.plot(
    aum_trend["date"],
    aum_trend["aum_crore"],
    marker="o"
)

plt.title("Industry Assets Under Management Trend")

plt.xlabel("Date")

plt.ylabel("AUM (Crore)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "aum_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 2
# Monthly SIP Inflow Trend
# ==========================================================

print("\nMonthly SIP Inflow Trend")

display(sip_df.head())

plt.figure(figsize=(10,5))

plt.plot(
    sip_df["month"],
    sip_df["sip_inflow_crore"],
    marker="o"
)

plt.title("Monthly SIP Inflow")

plt.xlabel("Month")

plt.ylabel("SIP Inflow (Crore)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "sip_inflow_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 3
# Top 5 Categories by Net Inflow
# ==========================================================

print("\nTop 5 Categories by Net Inflow")

top_category = (
    category_df
    .groupby("category")["net_inflow_crore"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

display(top_category)

plt.figure(figsize=(9,5))

plt.bar(
    top_category.index,
    top_category.values
)

plt.title("Top 5 Categories by Net Inflow")

plt.xlabel("Category")

plt.ylabel("Net Inflow (Crore)")

plt.xticks(rotation=20)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "top5_category_inflow.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 4
# Total Folio Trend
# ==========================================================

print("\nTotal Folio Trend")

display(folio_df.head())

plt.figure(figsize=(10,5))

plt.plot(
    folio_df["month"],
    folio_df["total_folios_crore"],
    marker="o"
)

plt.title("Total Mutual Fund Folios")

plt.xlabel("Month")

plt.ylabel("Total Folios (Crore)")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "folio_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ==========================================================
# Analysis 5
# Benchmark Index Trend
# ==========================================================

print("\nBenchmark Index Trend")

benchmark_trend = (
    benchmark_df
    .groupby("date")["close_value"]
    .mean()
    .reset_index()
)

display(benchmark_trend.head())

plt.figure(figsize=(10,5))

plt.plot(
    benchmark_trend["date"],
    benchmark_trend["close_value"]
)

plt.title("Average Benchmark Index Trend")

plt.xlabel("Date")

plt.ylabel("Average Close Value")

plt.xticks(rotation=45)

plt.tight_layout()

plt.savefig(
    FIGURE_PATH / "benchmark_trend.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

```
